```python
numeric_cols = [
    "CGPA", "Aptitude_Score", "Communication_Score",
    "Projects_Count", "Certifications_Count",
    "Skill_Count", "Package_LPA"
]

display(df[numeric_cols].describe().round(2))
```

```python
df["Placement_Status"] = df["Placed"].map({1: "Placed", 0: "Not Placed"}).fillna("Unknown")
df["Graduation_Year"] = pd.to_numeric(df["Graduation_Year"], errors="coerce")
df["CGPA"] = pd.to_numeric(df["CGPA"], errors="coerce")

STATUS_ORDER = ["Placed", "Not Placed", "Unknown"]
STATUS_COLORS = {
    "Placed": "#2ca02c",
    "Not Placed": "#d62728",
    "Unknown": "#9e9e9e"
}

print("Placement status counts:")
display(df["Placement_Status"].value_counts().reindex(STATUS_ORDER, fill_value=0).to_frame("Students"))
print("Graduation years:", sorted(df["Graduation_Year"].dropna().unique().tolist()))
```

```python
import pandas as pd
import plotly.express as px

# Load the placement dataset
df = pd.read_csv("placement_data_cleaned.csv")

# Create an interactive CGPA histogram
fig1 = px.histogram(
    df,
    x="CGPA",
    title="Distribution of Student CGPA",
    labels={"CGPA": "CGPA"}
)

fig1.update_layout(
    yaxis_title="Number of students",
    xaxis=dict(rangeslider=dict(visible=True, thickness=0.12)),
    bargap=0.05
)

fig1.show()
```

```python
import plotly.express as px

# Define the placement categories
STATUS_ORDER = ["Not Placed", "Placed"]

# Create an interactive scatter plot
fig2 = px.scatter(
    df,
    x="CGPA",
    y="Aptitude_Score",
    color="Placed",
    title="CGPA vs Aptitude Score by Placement Status",
    hover_data=[
        "Student_ID",
        "Branch",
        "Communication_Score",
        "Package_LPA"
    ],
    labels={
        "CGPA": "CGPA",
        "Aptitude_Score": "Aptitude Score",
        "Placed": "Placement Status"
    }
)

fig2.update_layout(
    xaxis_title="CGPA",
    yaxis_title="Aptitude Score"
)

fig2.show()
```

```python
print(df["Placed"].unique())
```

```python
placed_df = df[(df["Placed"] == 1) & (df["Package_LPA"] > 0)].copy()

fig3 = px.box(
    placed_df, x="Company_Type", y="Package_LPA",
    color="Company_Type",
    points="all",
    hover_name="Student_ID",
    hover_data={"Branch": True, "CGPA": True, "Skill_Count": True},
    labels={"Company_Type": "Company type", "Package_LPA": "Package (LPA)"},
    title="Package Distribution by Company Type (Placed Students)"
)
fig3.update_traces(marker=dict(size=5, opacity=0.65), jitter=0.35)
fig3.update_layout(showlegend=False)
fig3.show()
```

```python
# Summary values behind the box plot
if not placed_df.empty:
    display(placed_df.groupby("Company_Type")["Package_LPA"].agg(
        students="count", median_package="median", mean_package="mean"
    ).round(2))
else:
    print("No placed students with a positive package were found in this dataset.")
```

```python

import plotly.express as px

fig4 = px.scatter_3d(
    df,
    x="CGPA",
    y="Aptitude_Score",
    z="Communication_Score",
    color="Placed",
    title="3D Analysis of Student Placement Skills",
    hover_data=["Student_ID", "Branch", "Package_LPA"],
    labels={
        "CGPA": "CGPA",
        "Aptitude_Score": "Aptitude Score",
        "Communication_Score": "Communication Score",
        "Placed": "Placement Status"
    }
)

fig4.show()
```

```python

import plotly.express as px
import pandas as pd

# Create a copy for the animation
anim_df = df.copy()

# Convert graduation year to numeric and remove missing years
anim_df["Graduation_Year"] = pd.to_numeric(
    anim_df["Graduation_Year"], errors="coerce"
)
anim_df = anim_df.dropna(subset=["Graduation_Year"])

anim_df["Graduation_Year"] = anim_df["Graduation_Year"].astype(int)

# Sort graduation years
years = sorted(anim_df["Graduation_Year"].unique().tolist())

if len(years) > 0:
    fig5 = px.scatter(
        anim_df,
        x="CGPA",
        y="Aptitude_Score",
        color="Placed",
        animation_frame="Graduation_Year",
        hover_data=["Student_ID", "Branch"],
        title="Animated CGPA vs Aptitude Score by Graduation Year",
        labels={
            "CGPA": "CGPA",
            "Aptitude_Score": "Aptitude Score",
            "Placed": "Placement Status",
            "Graduation_Year": "Graduation Year"
        }
    )

    fig5.show()
else:
    print("No valid graduation years are available for animation.")
```
